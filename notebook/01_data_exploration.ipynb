# Superstore Sales & Business Intelligence Data Exploration
This notebook performs initial exploratory data analysis (EDA) on the Superstore dataset to uncover revenue patterns, profit margins, loss drivers, and customer segments.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned dataset
df = pd.read_csv('../data/clean/superstore_cleaned.csv')
    df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
print('Dataset shape:', df.shape)
```

## 1. High-Level Metrics Summary

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
overall_margin = (total_profit / total_sales) * 100
total_orders = df['Order ID'].nunique()
loss_orders = df[df['Is_Loss']].shape[0]

print(f'Total Sales: ${total_sales:,.2f}')
print(f'Total Profit: ${total_profit:,.2f}')
print(f'Profit Margin: {overall_margin:.2f}%')
print(f'Total Unique Orders: {total_orders:,}')
print(f'Loss-Making Transactions: {loss_orders:,} ({loss_orders/len(df)*100:.2f}%)')
```

## 2. Category & Sub-Category Performance Analysis

```python
category_perf = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Discount=('Discount', 'mean')
).reset_index()
category_perf['Profit_Margin_%'] = (category_perf['Total_Profit'] / category_perf['Total_Sales']) * 100
display(category_perf)
```

## 3. Impact of Discounts on Profitability

```python
discount_perf = df.groupby('Discount_Category', observed=False).agg(
    Orders=('Row ID', 'count'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Loss_Count=('Is_Loss', 'sum')
).reset_index()
discount_perf['Profit_Margin_%'] = (discount_perf['Total_Profit'] / discount_perf['Total_Sales']) * 100
display(discount_perf)
```
